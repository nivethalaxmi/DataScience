```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
```

```python
data=pd.read_csv("all_month.csv")
data
```

```python
df=data[['latitude','longitude']]
df
```

```python
df=df.dropna()
df
```

```python
model=DBSCAN(eps=2,min_samples=2)
labels=model.fit_predict(df)
labels
```

```python
plt.scatter(df['latitude'],df['longitude'],c=labels)
plt.show()
```

```python
df=pd.read_csv("data.txt.gz",sep=' ',names=column,nrows=20000)
df
```

```python
x=df[['temperature','humidity']].dropna()
model=KMedoids(n_clusters=3,random_state=0)
labels=model.fit_predict(x)

```

```python
import matplotlib.pyplot as plt
plt.scatter(x['temperature'],x['humidity'],c=labels ,cmap="viridis",s=5)
centers=model.cluster_centers_
plt.scatter(centers[:,0],centers[:,1],c="black",s=200,alpha=0.5)
plt.show()
```
