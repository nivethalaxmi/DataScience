```python
from scipy import stats
import numpy as np
data=[4.3,4.0,3.7,2,4.3,3.3,3.3,3.4,3.7,2.8]
ci= stats.t.interval(
    confidence =0.95,
    df=len(data) -1,
    loc=np.mean(data),
    scale=stats.sem(data)

)
```

```python
print("95% Confidence Interval :")

print("Lower;",ci[0])
print("Upper:",ci[1])
```

```python
from math import sqrt
data=[4.3,4.0,3.7,3.8,3.4,2.8,3.4,3.8,4.3,2.8,
      4.4,41,4,4,1,3.5,3.9,3.4,3.7,3.5,4.1,4.2,
       4,4,3.6,2.9,3.9,3.7,3.3,3.1,3.6,2.8,4.3
      ]
std=3
ci=stats.norm.interval(
    0.95,
    loc=np.mean(data),
    scale=std/np.sqrt(len(data))
)
print("Number of data:",len(data))
print("mean:",np.mean(data))
print("95% Confidence Interval:")
print("Lower:",ci[0])
print("Upper:",ci[1])
```
