```python
import pandas as pd
df=pd.read_csv("2015.csv")
df
```

```python
df=df["Family"]
df
```

```python
from scipy.stats import skew
result=skew(df)
print(df)
if result >0:
  print("Positive skew--right side long tail")
elif result<0:
  print("negative skew--left side long tail")
else:
  print("symmetric skew")
```

```python
import matplotlib.pyplot as plt
plt.hist(df,color="blue",edgecolor="black")
plt.xlabel("Famil Count")
plt.title("Population distrubution")
plt.show()
```

```python
df=pd.read_csv("Microsoft_Stock.csv")

df
```

```python
df=df["Volume"]
df
```

```python
answer=skew(df)
print(answer)
if answer>0:
  print("Postive skew")
elif answer<0:
  print("Negative skew")
else:
  print("symmetric skew")
```
