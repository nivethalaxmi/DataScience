```python
#liner_regression
import pandas as pd

```

```python
data=pd.read_csv("Ice Cream Sales - temperatures.csv")
data
```

```python
data.isnull().sum()
```

```python
data.dropna(inplace=True)
data
```

```python
data.drop_duplicates(inplace=True)
data
```

```python
data.describe()
```

```python
data.info()
```

```python
x=data['Temperature']
x
```

```python
y=data['Ice Cream Profits']
y
```

```python
#package
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
x_train,x_text,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=30)

```

```python
#training the model
model=LinearRegression()
model.fit(x_train.values.reshape(-1, 1), y_train)
```

```python
#validation
y_pred=model.predict(x_text.values.reshape(-1,1))
acc=r2_score(y_test,y_pred)
print(acc*100)
```

```python
output=int(input("Enter the temp"))
input=[[output]]
print("ice:",model.predict(input))
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y,color="Blue",label="data")
plt.plot(x_text,y_pred,color="red",label="model")
plt.title("Ice Cream Sales - temperatures")
plt.xlabel("temperature")
plt.ylabel("Ice Cream")
plt.legend()
plt.show()
```
