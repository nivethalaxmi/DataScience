```python
#MultiLinear_regression

import pandas as pd
import numpy as np
data=pd.read_csv("Advertising.csv")
data

```

```python
data.isnull().sum()
```

```python
x1=data['Newspaper']
x1
```

```python
x2=data['Radio']
x2
```

```python
x3=data['TV']
x3
```

```python
x=np.column_stack(( x1,x2,x3))
x
```

```python
y=data['Sales']
y
y=np.array(y).reshape(-1,1)
y
```

```python

from sklearn.model_selection import train_test_split as tain_test_split
x_train,x_text,y_train,y_test=tain_test_split(
    x,y,test_size=0.40,random_state=42
)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x,y)
```

```python
news_paper=float(input("Enter the Newspaper"))
Ra_dio=float(input("Enter the Radio"))
T_v=float(input("Enter the tv"))
inputs=np.column_stack((news_paper,Ra_dio,T_v))
print(model.predict(inputs))
```

```python
from sklearn.metrics import r2_score
x_test=x
y_test=y
y_pred=model.predict(x_test)
acc=r2_score(y_test,y_pred)
print(round(acc*100,2))
```
