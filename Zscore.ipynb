```python
#Bank gets 3 customers/5 min; probability 5 in next 5 min.
from scipy.stats import poisson as p
import matplotlib.pyplot as plt
import numpy as np
k=3
lam=5
df=p.pmf(k,lam)
print(df)
```

```python
n=np.arange(1,10)
print(n)
```

```python
dis=p.pmf(n,lam)
print(dis)
```

```python
plt.bar(n,dis)
plt.xlabel('Number of customers')
plt.ylabel('Probability')
plt.show()
```

```python
# Website gets 6 visits/min; probability exactly 10 next minute.
k=10
lam=6
df=p.pmf(k,lam)
print(df)
```

```python
out=np.arange(1,13)
print(out)
```

```python
dis=p.pmf(out,lam)
print(dis)
```
